```python
import torch
from torch.utils.data import DataLoader
from torchvision import transforms

from src.dataloader.dataset import SecondStitchDataset

# ImageNet transforms (required for ResNet)
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

dataset = SecondStitchDataset(
    "data/processed/train",
    transform=transform
)

dataloader = DataLoader(
    dataset,
    batch_size=16,
    shuffle=True,
    num_workers=2
)

# Test one batch (NO style)
images, category, occlusion = next(iter(dataloader))

print("Images shape:", images.shape)        # [16, 3, 224, 224]
print("Category shape:", category.shape)    # [16]
print("Occlusion shape:", occlusion.shape)  # [16]

print("Category min/max:", category.min().item(), category.max().item())
print("Occlusion min/max:", occlusion.min().item(), occlusion.max().item())
```

```python
import torch
import torch.nn as nn
from torch.optim import Adam

from src.models.resnet18_multitask import ResNet18MultiTask

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# Model (2 heads: category + occlusion)
model = ResNet18MultiTask().to(device)
model.train()

criterion_category = nn.CrossEntropyLoss()
criterion_occlusion = nn.CrossEntropyLoss()

optimizer = Adam(model.parameters(), lr=1e-4)

num_epochs = 21

for epoch in range(num_epochs):
    running_loss = 0.0
    batches = 0

    for images, category, occlusion in dataloader:
        images = images.to(device)
        category = category.to(device)
        occlusion = occlusion.to(device)

        optimizer.zero_grad()

        # Forward (ONLY TWO OUTPUTS NOW)
        category_logits, occlusion_logits = model(images)

        loss_cat = criterion_category(category_logits, category)
        loss_occ = criterion_occlusion(occlusion_logits, occlusion)

        loss = loss_cat + loss_occ
        loss.backward()
        optimizer.step()

        running_loss += loss.item()
        batches += 1

        if batches >= 200:
            break

    print(f"Epoch [{epoch+1}/{num_epochs}] - Avg Loss: {running_loss / batches:.4f}")
```

```python
CATEGORY_MAP = {
    0: "short_sleeve_top",
    1: "long_sleeve_top",
    2: "short_sleeve_outwear",
    3: "long_sleeve_outwear",
    4: "vest",
    5: "sling",
    6: "shorts",
    7: "trousers",
    8: "skirt",
    9: "short_sleeve_dress",
    10: "long_sleeve_dress",
    11: "vest_dress",
    12: "sling_dress"
}
```

```python
LONG_TOPS = {
    "long_sleeve_top",
    "short_sleeve_top",
    "long_sleeve_outwear",
    "short_sleeve_outwear"
}

BOTTOMS_LONG = {
    "trousers"
}

DRESSES = {
    "short_sleeve_dress",
    "long_sleeve_dress",
    "vest_dress",
    "sling_dress"
}
```

```python
def generate_upcycling_ideas(category_id, occlusion_id=None):
    """
    category_id: int (1–13 based on CATEGORY_MAP)
    occlusion_id: optional (0=visible, 1=partial, 2=heavy)
    """

    category = CATEGORY_MAP.get(category_id, "unknown")
    ideas = []

    # Rule 1: Long shirt / top → Crop top / Sleeveless
    if category in LONG_TOPS:
        ideas.append("Convert into a crop top")
        ideas.append("Redesign as a sleeveless top")

    # Rule 2: Jeans / long pant → Skirt / Shorts
    if category in BOTTOMS_LONG:
        ideas.append("Reconstruct into a skirt")
        ideas.append("Convert into shorts")

    # Dresses → Top / Tunic / Sleeveless variants
    if category in DRESSES:
        ideas.append("Convert into a sleeveless dress")
        ideas.append("Shorten length to create a tunic-style top")

    # Occlusion-aware refinement (optional but good)
    if occlusion_id == 1:
        ideas.append("Use contrast fabric panels to replace missing sections")

    if occlusion_id == 2:
        ideas.append("Apply patchwork reconstruction using recovered fabric")

    # Safety fallback
    if not ideas:
        ideas.append("Modify garment using modular upcycling techniques")

    return ideas
```
